# Comparing GTM and SING

Graphical transformation models (GTMs) and Sparsity Identification in Non-Gaussian distributions (SING) address undirected graph recovery under nonlinear dependence. This report compares their parameter counts and the derivative calculations associated with identification of the Markov Random Field / Undirected Graphical Model (UGM).

For a positive, twice continuously differentiable density $p$, both approaches characterise conditional independence by the hessian w.r.t. the data:

$$X_i\perp X_j\mid X_{-ij}
\quad\Longleftrightarrow\quad
\partial_i\partial_j\log p(x)=0\ \text{for all }x.$$

### Metrics

SING estimates an integrated score from the data and squares it:

$$\widehat\Omega_{ij}=\frac{1}{N}\sum_{n=1}^{N}
\left[\partial_i\partial_j\log\widehat p(x_n)\right]^2.$$

GTM diagnostics are evaluated on model-generated samples to assess dependence across the fitted distribution:
$$
\oabbfHtuv
		= \int f(\bftz)\, |\bbfHtuv(\bftz)| \,\mathrm{d}\bftz
		\approx \frac{1}{S}\sum_{s=1}^{S} |\bbfHtuv(\bftz)| ,
$$
$$
	\iae\nolimits_{u,v|-u,v}
	= \frac{1}{2}\int f(\bftz_{-u,v})
	\left(\int|\fuv - \fuvind|\,\mathrm{d}\bftz_{u,v}\right)
	\mathrm{d}\bftz_{-u,v} \\
    \approx \frac{1}{2S}\sum_{s=1}^{S}
	\frac{f(\bftz_{-u,v}^s)}{f(\bftz^s)}
	\times|\fuvs - \fuvinds|,
$$

##### **1. Difference: GTM uses synthetic samples to evaluate the fitted distribution while SING merely looks at the samples.However the GTM approach can be adapted to SING as well.** 
##### **2. Difference: GTM also offers a likelihood based interpretable metric in the $\iae$. However this can also be adapted to SING as well.** 

### Parameter counts

Let $D$ denote data dimensionality, $N$ the number of observations, $L$ the number of GTM dependence layers, and $B_{\mathrm{GTM}}$ the number of coefficients per GTM univariate transformation. For SING, $B$ denotes the polynomial-degree index. The numerical illustration uses $B_{\mathrm{GTM}}=10$ and $B\in\{2,3\}$ for SING.
For the GTM we have $DB_{\mathrm{GTM}}$ transformation layer and $L\frac{D(D-1)}{2}B_{\mathrm{GTM}}$ decorrelation layer parameters and thus in total:
$$P_{\mathrm{GTM}}=DB_{\mathrm{GTM}}+L\frac{D(D-1)}{2}B_{\mathrm{GTM}}$$
For SING we have 
$$P_{\mathrm{SING}}=\binom{D+B+1}{B+1}-1.$$
For SING which uses a dense triangular transport in \(D\) dimensions, \(c_k\) depends on \(k-1\) variables and uses total degree at most \(B\), while \(h_k\) depends on \(k\) variables and uses total degree at most \(B-1\). The corresponding full index-set cardinalities give
$$
P_{c_k}=\binom{k+B-1}{B},
\qquad
P_{h_k}=\binom{k+B-1}{B-1}.
$$
Summing over the \(D\) transport components yields
$$
P_c=\sum_{k=1}^{D}P_{c_k}=\binom{D+B}{B+1},
\qquad
P_h=\sum_{k=1}^{D}P_{h_k}=\binom{D+B}{B}-1.
$$
Therefore, the combined coefficient count is
$$
P_{\mathrm{SING}}=P_c+P_h
=\binom{D+B+1}{B+1}-1.
$$
This is based on SING section 3.3. with formulas (14) and (15) as well as the cited work Spantini et al. (2018).
This exact count is never established in the paper, however we test it on ther code for different $B,D$ combinations and it is correct.


##### **3. Difference:  At fixed $B$ and $L$, $P_{\mathrm{GTM}}=O(D^2)$, whereas $P_{\mathrm{SING}}=O(D^{B+1})$. The latter reflects the rapid growth of the multivariate tensor product basis. Equal numerical values of $B$ do not imply equal flexibility or approximation error across the two model classes, typically $B=10$ for the GTM while $B=2/3$ for SING. Nevertheless already with $B=2$ we have cubic compared to sqaure growth of parameters with dimensionality $P_{\mathrm{SING}}=O(D^{3}) > P_{\mathrm{GTM}}=O(D^2)$.**




### Derivatives required to Compute Hessian Metrics

GTM and SING both need $D^2$ computations for the Hessian metrics of one sample.
SING's variance approximation howwever is the real bottleneck of SING.
It uses the score gradient $g_{ij}=\nabla_\theta\widehat\Omega_{ij}$ and estimated Fisher information,

$$\widehat F=-\frac{1}{N}\sum_{n=1}^{N}
\nabla_\theta^2\log\widehat p_\theta(x_n),
\qquad
\widehat{\mathrm{Var}}(\widehat\Omega_{ij})
\approx\frac{1}{N}g_{ij}^{\mathsf T}\widehat F^{-1}g_{ij}.$$

Here $\theta$ contains $P_{\mathrm{SING}}$ coefficients. Details are in SING 4.1 and 4.2 where (23) is the variance estimator that is so expensiv. Evaluating all entries and assuming two SING iterations yields the derivative-count proxy

$$C_{\mathrm{SING}}=2\left(
\underbrace{ND^2}_{\text{data Hessians}}+
\underbrace{ND^2P_{\mathrm{SING}}}_{\text{mixed data/parameter derivatives}}+
\underbrace{NP_{\mathrm{SING}}^2}_{\text{parameter Hessians}}\right).$$

The factor of two accounts for two SING iterations, approximating both with the same dense coefficient count. This is an assumption for the comparison rather than a fixed iteration requirement of the algorithm; the constant factor does not change the asymptotic dimensional scaling.

In comparison we have the GTM Bootstrap approach to uncertainty. Let $R$ be the number of refits and the approximate model-evaluation cost by $I*P_{GTM}$ which contains the number of training iterations needed in each retraining and the number of parameters. Hence we have $R*I*P_{GTM}*N$ retraining costs and $R*ND^2$ the number of derivatives for evaluation. So in total we have:

$$C_{\mathrm{GTM}}=R*I*P_{GTM}*N + R*ND^2.$$

Holding $B,L,I,R,N$ fixed and using $P_{\mathrm{GTM}}=O(D^2)$ and the dense expansion count $P_{\mathrm{SING}}=O(D^{B+1})$, the proposed proxies yield
$$
C_{\mathrm{GTM}}=O(D^2),
$$
whereas
$$
C_{\mathrm{SING}}
=
O\!\left(D^2+D^{B+3}+D^{2B+2}\right)
=
O(D^{2B+2}).
$$

##### **4. Difference:  At fixed $B,L,I,R,N$, $C_{\mathrm{GTM}}=O(D^2)$, whereas $C_{\mathrm{SING}}=O(D^{2B+2})$. Thus, for SING degrees $B=2$ and $B=3$, the dense proxy grows as $O(D^6)$ and $O(D^8)$, respectively, compared with $O(D^2)$ for the proposed GTM. This explains why running SING in larger then 15 dimensions as was done in the paper (6.5) gets expensiv very quickly.**

### Applicability in Large D and Small N

An additional Problem of Sing is that the Variance estimators require the inverse Fischer Information. See Formula (23) where it is used in the Variance Estimator and Formula (19) where the Fischer Information is defined as the Expected Hessian, e.g. computed using average over dataset.
For large D and moderate B we have very many parameters $P_{SING}$ and if those are more then N so $P_{SING}>N$ then this is a warnng sign that the estiamted Fischer Information might not be positive semidefinite but rather singular. This is not necessarily the case but the risk increases. Flat dimensions in the Optimsiation space, thus small eigenvalues, will increase the variance estimator. They are parameter directions barely constrained by the data. Further only under regularity conditions and enough information we have that the curvature describes how the estimator would vary across repeated datasets. But with a highly flexible model relative to the available observations, that approximation can be poor.
SINGs iterative approach which makes many parameters close to zero can help.

Briefly said: SING’s variance estimator requires reliably estimated, invertible curvature and suitable asymptotic assumptions. Large transport expansions can challenge both requirements.

## Interpretation

- **Theoretical guarantees.** SING provides an asymptotic graph-recovery consistency result for an idealised single-step estimator under stated assumptions [1, Section 4.4]. Bootstrap assessment for GTM does not by itself establish comparable graph-selection guarantees; calibrated $p$-values and theoretical error control remain directions for further development.

- **Uncertainty assessment.** SING uses a parameter-level variance approximation to construct score thresholds. GTM can assess uncertainty through bootstrap refitting, optionally retaining the original tuning settings. Additional GTM metrics, particularly the $\iae$, are more readily interpretable, and GTM's sample-based evaluation assesses the learned distribution more broadly rather than being restricted to the training observations.

- **Parameter identifiability.** GTM is a non-identifiable neural network, whereas SING requires an identifiable transformation. The combination of identifiability and the need for sufficiently general transformations to limit the influence of variable ordering probably explains the use of a tensor-product basis. Thus, thresholding comes at the cost of a larger parameterisation, in terms of the number of parameters.

- **Representation and ordering.** SING uses a monotone triangular transport map and iteratively updates its ordering and sparsity. This ordering defines a statistical factorisation, not a causal DAG. GTM uses a layered transformation architecture without a separate ordering-search procedure.

- **Interpretation of dependence.** SING's nonnegative, integrated Hessian score summarises conditional-dependence magnitude. GTM complements its aggregate diagnostics with signed local conditional pseudo-correlations, describing how conditional associations vary across the fitted distribution.

- **Regularisation and generalisation.** GTM explicitly penalises transformation roughness and conditional dependence. These penalties provide mechanisms for controlling overfitting; any resulting advantage in held-out log likelihood requires empirical evaluation.

```python
from math import comb
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

B_GTM = 10
B_SING = [2, 3]
SING_ITERATIONS = 2  # apply the dense work proxy to each iteration
L = 3
N = 2000
R = 100  # bootstrap repetitions
I = 2000  # training iterations per bootstrap refit
D_MIN = 1
D_MAX = 100
LOG_Y = True

for name, value in {'B_GTM': B_GTM, 'L': L, 'N': N, 'R': R, 'I': I, 'SING_ITERATIONS': SING_ITERATIONS,
                    'D_MIN': D_MIN, 'D_MAX': D_MAX}.items():
    if isinstance(value, bool) or not isinstance(value, int) or value < 1:
        raise ValueError(f'{name} must be a positive integer')
if not B_SING or any(isinstance(b, bool) or not isinstance(b, int) or b < 1 for b in B_SING):
    raise ValueError('B_SING must contain positive integer degrees')
if len(set(B_SING)) != len(B_SING):
    raise ValueError('B_SING must contain distinct degrees')
if D_MAX < D_MIN:
    raise ValueError('D_MAX must be at least D_MIN')
```

```python
dimensions = list(range(D_MIN, D_MAX + 1))
# Use separate basis settings for GTM and each SING curve.
p_gtm = [d * B_GTM + L * (d * (d - 1) // 2) * B_GTM for d in dimensions]
p_sing = {b: [comb(d + b + 1, b + 1) - 1 for d in dimensions] for b in B_SING}

# GTM proxy from the text: bootstrap training work plus derivative evaluation.
gtm_training_work = [R * I * p * N for p in p_gtm]
gtm_derivative_work = [R * N * d ** 2 for d in dimensions]
c_gtm = [training + evaluation for training, evaluation in zip(
    gtm_training_work, gtm_derivative_work)]

# SING: data Hessians, mixed derivatives, and parameter Hessians.
sing_data_hessian = [N * d ** 2 for d in dimensions]
sing_mixed_derivatives = {
    b: [N * d ** 2 * p for d, p in zip(dimensions, p_sing[b])] for b in B_SING}
sing_parameter_hessian = {b: [N * p ** 2 for p in p_sing[b]] for b in B_SING}
c_sing_per_iteration = {
    b: [data + mixed + parameter for data, mixed, parameter in zip(
        sing_data_hessian, sing_mixed_derivatives[b], sing_parameter_hessian[b])]
    for b in B_SING}
c_sing = {b: [SING_ITERATIONS * value for value in c_sing_per_iteration[b]]
          for b in B_SING}

count_columns = {
    'GTM parameters': p_gtm,
    'GTM training work': gtm_training_work,
    'GTM derivative work': gtm_derivative_work,
    'GTM total': c_gtm,
    'SING data Hessians (per iteration)': sing_data_hessian,
}
for b in B_SING:
    count_columns.update({
        f'SING B={b} parameters': p_sing[b],
        f'SING B={b} mixed derivatives (per iteration)': sing_mixed_derivatives[b],
        f'SING B={b} parameter Hessians (per iteration)': sing_parameter_hessian[b],
        f'SING B={b} per-iteration total': c_sing_per_iteration[b],
        f'SING B={b} total': c_sing[b],
    })
# Retain exact Python integer counts for every plotted dimension.
counts = pd.DataFrame(count_columns, index=pd.Index(dimensions, name='D'), dtype=object)
```

```python
def comparison_plot(gtm_values, sing_values_by_degree, title, ylabel):
    fig, ax = plt.subplots(figsize=(7.0, 3.5), constrained_layout=True)
    ax.plot(dimensions, np.asarray(gtm_values, dtype=float),
            label=rf'GTM ($B={B_GTM}$)', color='#2563eb', linewidth=2.5)
    colors = ['#ea580c', '#15803d', '#9333ea', '#be123c']
    styles = ['--', '-.', ':', '--']
    for index, b in enumerate(B_SING):
        ax.plot(dimensions, np.asarray(sing_values_by_degree[b], dtype=float),
                label=rf'SING ($B={b}$)', color=colors[index % len(colors)],
                linestyle=styles[index % len(styles)], linewidth=2.5)
    if LOG_Y:
        ax.set_yscale('log')
    ax.set(xlabel='Data dimensionality D', ylabel=ylabel, title=title)
    if D_MIN < D_MAX:
        ax.set_xlim(D_MIN, D_MAX)
    ax.grid(True, which='major', alpha=0.25)
    ax.legend(frameon=False)
    return fig
```

## Proposed discussion paragraph (draft 1)

GTM and SING recover undirected conditional-dependence structures in nonlinear, non-Gaussian distributions using derivatives of the fitted log density. SING provides asymptotic variance-based thresholding under identifiability and regularity assumptions [1], whereas GTM has a non-identifiable transformation and therefor assesses uncertainty through bootstrap refitting. At fixed basis sizes and layer number, $P_{\mathrm{GTM}}=O(D^2)$ and $P_{\mathrm{SING}}=O(D^{B+1})$, giving cubic or quartic SING parameter growth for $B=2$ or $B=3$. This is due to SING using generic identifiable tensor product interactions in its triangular transport map to mitigate the effect of initial ordering. Furthermore the uncertainty assessement in SING relies on asymptotic variance estimates that require $C_{\mathrm{SING}}=ND^2+ND^2P_{\mathrm{SING}}+NP_{\mathrm{SING}}^2=O(D^{2B+2})$ derivatives to compute, hence $O(D^6)$ or $O(D^8)$, compared with $C_{\mathrm{GTM}}=R*I*P_{GTM}*N + R*ND^2=O(D^2)$ for a GTM with a fixed sample size $N$, training iterations $I$, and bootstrap repetitions $B$. This highlights a trade-off between identifiability-based asymptotic inference and a potentially more scalable, non-identifiable representation with bootstrap uncertainty.  SING iteratively reorders and sparsifies its transport map and controls flexibility through polynomial degree, whereas GTM selects smoothness and sparsity penalties through tuning. Compared to SING, we additionally propose evaluation on model-generated samples which look at the estimated distribution as oppossed to SING where the training sample points are evaluated. Furthermore we propose an interpretable likelihood-based metric with the $\iae$, and the more easily interpretable signed local conditional pseudo-correlations can be used to understand the nonlinear dependency structure. Sampling based evaluation and other metrics except the local pseudo correlations can also be applied in the SING framework, although they further increase computational cost.
